```python
import os
print(os.getenv("TAVILY_API_KEY"))
```

```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from tavily import TavilyClient
```

```python
@tool
def search(query: str) -> str:
    """Simulate a weather search for a given query 
    and return a  predefined weather report for Tehran."""

    print(f"Searching for the weather conditions")
    return "Tehran weather is sunny."
```

```python
print(search.name)
print(search.description)
print(search.args)
```

```python

result = search.invoke(
    {
        "query": "Weather in Tehran"
    }
)
print(result)
```

```python
llm = ChatOpenAI(
    model="nvidia/nemotron-3.5-super-vl:free",
    api_key=os.getenv("OpenRouter_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)
tools = [search]
```

```python
agent = create_agent (
    model =llm,
    tools = tools
)
```

```python
result = agent.invoke({
    'messages': [HumanMessage (content="What is the weather in Tehran?")]
})
```

```python
type(result)
```

```python
for i, message in enumerate(result["messages"]):
    print(f"\n{'='*50}")
    print(f"Message {i + 1}: {type(message).__name__}")
    print(f"{'='*50}")

    print(message.model_dump)
```

```python
@tool
def real_search(query:str) -> str:
    """
    Tool that searches over the internet
    Args:
        query: The query to search for
    Returns:
        The search results
    """
    print(f'Searching for the {query}')
    return(tavliy.search(query=query,
                         max_results=5))
```

```python
llm = ChatOpenAI(
    model="nvidia/nemotron-3.5-super-vl:free",
    api_key=os.getenv("OpenRouter_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)
```

```python
tools = [real_search]
agent = create_agent(
    model=llm,
    tools=tools
)
```

```python
result = agent.invoke({
    'messages': [HumanMessage (content="What is the weather in Tehran?")]
})
```
